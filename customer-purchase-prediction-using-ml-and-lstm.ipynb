# Predict Buying Intention

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
import xgboost as xgb

```

```python
X_train = pd.read_csv('/kaggle/input/x-and-y-test/X_test.csv')
X_test = pd.read_csv('/kaggle/input/x-and-y-test/X_train.csv')
y_train = pd.read_csv('/kaggle/input/x-and-y-test/y_test.csv')
y_test = pd.read_csv('/kaggle/input/x-and-y-test/y_train.csv')
```

# Logistic Regression

```python
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train.values.ravel())
y_pred_log = log_model.predict(X_test)
print("Logistic Regression:\n", classification_report(y_test, y_pred_log))
```

# XGBoost

```python
xgb_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train, y_train.values.ravel())
y_pred_xgb = xgb_model.predict(X_test)
print("XGBoost:\n", classification_report(y_test, y_pred_xgb))
```

 # Survival Analysis (Cox Regression)

```python
import pandas as pd
from lifelines import CoxPHFitter

# Load dataset
df = pd.read_csv('/kaggle/input/survival-analysis-dataset/survival_analysis_dataset.csv')

# Encode categorical
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['location'] = df['location'].astype('category').cat.codes

# Drop rows where time_until_purchase is NaN
df = df.dropna(subset=['time_until_purchase'])

# Fit Cox Model
cph = CoxPHFitter()
cph.fit(df[[
    'age', 'gender', 'location', 'website_visits_last_30_days',
    'avg_time_on_site_minutes', 'clickstream_depth', 'total_purchases',
    'days_since_last_purchase', 'email_opened_last_30_days',
    'add_to_cart_count', 'product_reviews_count', 'product_view_count',
    'avg_time_per_product_min', 'time_until_purchase', 'event'
]], duration_col='time_until_purchase', event_col='event')

cph.print_summary()
```

# Recurrent Neural Network (RNN or LSTM) – Time to Purchase

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout, Input


# Load and preprocess
df = pd.read_csv('/kaggle/input/survival-analysis-dataset/survival_analysis_dataset.csv')
df = df.dropna(subset=['time_until_purchase'])

# Encode categoricals
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['location'] = LabelEncoder().fit_transform(df['location'])

# Feature selection
features = [
    'age', 'gender', 'location', 'website_visits_last_30_days',
    'avg_time_on_site_minutes', 'clickstream_depth', 'total_purchases',
    'days_since_last_purchase', 'email_opened_last_30_days',
    'add_to_cart_count', 'product_reviews_count', 'product_view_count',
    'avg_time_per_product_min'
]

X = df[features]
y = df['time_until_purchase']

# Scale features
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Reshape for LSTM (samples, time_steps, features)
X_lstm = np.reshape(X_scaled, (X_scaled.shape[0], 1, X_scaled.shape[1]))

# Build LSTM model (cleaner version)
model = Sequential()
model.add(Input(shape=(X_lstm.shape[1], X_lstm.shape[2])))  # Cleaner input
model.add(LSTM(units=64, return_sequences=False))
model.add(Dropout(0.2))
model.add(Dense(1))  # Predict time

model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(X_lstm, y, epochs=20, batch_size=32)

# Predict
predicted_time = model.predict(X_lstm)

```

# Train LSTM on Top 1000 Customers

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout, Input

# Load dataset
df = pd.read_csv('/kaggle/input/survival-analysis-dataset/survival_analysis_dataset.csv')
df = df.dropna(subset=['time_until_purchase'])

# Encode categorical columns
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['location'] = LabelEncoder().fit_transform(df['location'])

# Feature selection
features = [
    'age', 'gender', 'location', 'website_visits_last_30_days',
    'avg_time_on_site_minutes', 'clickstream_depth', 'total_purchases',
    'days_since_last_purchase', 'email_opened_last_30_days',
    'add_to_cart_count', 'product_reviews_count', 'product_view_count',
    'avg_time_per_product_min'
]

X = df[features]
y = df['time_until_purchase']

# ✅ Keep only the top 1000 rows
X = X.iloc[:1000]
y = y.iloc[:1000]

# Normalize
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Reshape for LSTM
X_lstm = np.reshape(X_scaled, (X_scaled.shape[0], 1, X_scaled.shape[1]))

# Build and train model
model = Sequential()
model.add(Input(shape=(X_lstm.shape[1], X_lstm.shape[2])))
model.add(LSTM(units=64, return_sequences=False))
model.add(Dropout(0.2))
model.add(Dense(1))

model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(X_lstm, y, epochs=20, batch_size=32)

# Predict
predicted_time = model.predict(X_lstm)
```

# Key Finings

In this study, we examined various machine learning models to predict customer buying intentions and the expected time of purchase. Logistic Regression achieved an accuracy of **64%** but ***performed poorly*** in identifying actual buyers, with a recall rate of just **1%**. This highlights its ***limitations in handling imbalanced data.***

XGBoost outperformed Logistic regression capturing more non-linear patterns and improving the buyer recall rate to **21%**, making it a more suitable choice for predicting purchase intent. For predicting the time to purchase, Cox Regression yielded a **low concordance** score of **0.52**, indicating **weak predictive power** and **statistically insignificant feature contributions**. 

In contrast, the **LSTM** model demonstrated strong performance, significantly reducing loss during training and effectively modeling complex behavioral sequences. Overall, **XGBoost** is the best option for predicting who will buy, while **LSTM** is the most effective for predicting when purchases will occur. Traditional models like Logistic and Cox Regression are less effective for this dynamic behavioral dataset.

# Conclusion

This analysis demonstrates that customer buying behavior is better captured using advanced, non-linear models. While traditional methods like Logistic and Cox Regression fall short, models such as XGBoost and LSTM effectively handle complex patterns and temporal dynamics. Therefore, for real-world applications, we recommend using XGBoost to predict buying intention and LSTM to estimate when a customer is likely to make a purchase.
